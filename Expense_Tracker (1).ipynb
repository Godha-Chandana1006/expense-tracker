```python
import csv
import os
from datetime import date
FILE_NAME = "/content/drive/MyDrive/expenses.csv"
class Expense:
  def __init__(self,amount,category,description,expense_date=""):
    self.amount=amount
    self.category=category
    self.description=description
    self.expense_date=expense_date
def save_expenses(expenses):
  try:
    with open(FILE_NAME,"w",newline="",encoding="utf-8") as file:
       fields=["amount","category","description","expense_date"]
       writer=csv.DictWriter(file,fieldnames=fields)
       writer.writeheader()
       for expense in expenses:
         writer.writerow({"amount":f"{expense.amount:.2f}",
                          "category":expense.category,
                          "description":expense.description,
                          "expense_date":expense.expense_date})
  except(OSError,csv.Error) as error:
    print("Could not save expenses",error)
def load_expenses():
  expenses=[]
  if not os.path.exists(FILE_NAME):
    return expenses
  try:
    with open(FILE_NAME,"r",newline="",encoding="utf-8") as file:
      reader=csv.DictReader(file)
      for row in reader:
        expense=Expense(float(row["amount"]),
                        row["category"],
                        row["description"],
                        row.get("expense_date",""))
        expenses.append(expense)
  except (ValueError,KeyError,TypeError) as error:
     print("Some saved data could not be read:",error)
  except(OSError,csv.Error) as error:
    print("Could not read the expense file",error)
  return expenses
def add_expense(expenses):
  category=input("Enter expense category:").strip().title()
  if not category:
    print("Category cannot be empty")
    return
  description=input("Enter expense description:").strip()
  if not description:
    print("Description cannot be empty")
    return
  expense_date = input("Enter date (YYYY-MM-DD): ").strip()
  if not expense_date:
    expense_date=str(date.today())
    print(f"No date entered. Using today's date: {expense_date}")
  try:
    amount=float(input("Enter amount in Rs:"))
    if amount<=0:
      print("Amount must be graeter than zero")
      return
  except ValueError:
    print("Invalid input.Please enter a number")
    return
  expense=Expense(amount,category,description,expense_date)
  expenses.append(expense)
  save_expenses(expenses)
  print("Expense added and saved successfully")
def view_expenses(expenses):
  if not expenses:
    print("No expenses recorded yet")
    return
  print("\n----All Expenses----")
  for number,expense in enumerate(expenses,start=1):
    print(f"{number}.{expense.category}-"
    f"{expense.amount:.2f}-"
    f"{expense.description}-"
    f"{expense.expense_date}")
def summarize_expenses(expenses):
  if not expenses:
    print("No expenses to summarize")
    return
  total=0
  category_totals={}
  for expense in expenses:
    total+=expense.amount
    if expense.category in category_totals:
      category_totals[expense.category]+=expense.amount
    else:
      category_totals[expense.category]=expense.amount
  print("\n---Expense Summary")
  print(f"Total Spending:Rs.{total:.2f}")
  print(f"Number of expenses:{len(expenses)}")
  print("\nSpending by category")
  for category,amount in category_totals.items():
    print(f"{category}:Rs.{amount:.2f}")
def main():
  expenses=load_expenses()
  print(f"Loaded {len(expenses)} saved expense")
  while True:
    print("\n-----EXPENSE TRACKER-----")
    print("1.Add expense")
    print("2.View expenses")
    print("3.Summmarize expenses")
    print("4.Exit")
    choice=input("Choose an option(1-4):").strip()
    if choice=="1":
      add_expense(expenses)
    elif choice=="2":
      view_expenses(expenses)
    elif choice=="3":
      summarize_expenses(expenses)
    elif choice=="4":
      print("Thank you for using Expense Tracker")
      break
    else:
      print("Invalid choice.Please select 1,2,3 or 4")
main()
```
